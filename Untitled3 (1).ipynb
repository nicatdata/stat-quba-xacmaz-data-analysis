```python
import pandas as pd
import numpy as np

file_path = '008 (4).xlsx'
# 26-cı sətir real başlıq olduğu üçün header=26 təyin edirik
df = pd.read_excel(file_path, header=26, sheet_name='QUBA-XAÇMAZ')

print("Fayl uğurla oxundu! Ölçü:", df.shape)
print(df.head(3))
```

Bu addımda Excel faylını əvvəl tapdığımız 26-cı sətirdən başlayaraq (header=26) oxuduq. Beləliklə, sütun adları və başlıqlar artıq yerinə oturdu.

```python
# "Unnamed" ilə başlayan avtomatik yaranan boş sütunları silirik
df = df.loc[:, ~df.columns.astype(str).str.startswith("Unnamed")]

# İlk sütunu region olaraq adlandırırıq
df = df.rename(columns={df.columns[0]: "region"})
df = df.dropna(subset=["region"])

# Şərti işarələri (məsələn, "-", "...") Python-un başa düşəcəyi NaN-a çevririk
df = df.replace({"-": np.nan, "...": np.nan, "x": np.nan, "X": np.nan})

print("Təmizlənmə tamamlandı! Qalan sütunlar:", df.columns[:5])
```

Excel-dən gələn lazımsız Unnamed sütunlarını sildik, birinci sütunu region adlandırdıq və cədvəldəki -, ... kimi şərti simvolları analizi poza bilmə deyə NaN (boş dəyər) formatına çevirdik.

```python
# İllərin sütun şəklində olmasını (wide format) long formata çevririk
long_df = df.melt(id_vars="region", var_name="year", value_name="value")

print("Long format uğurla yaradıldı! Yeni data ölçüsü:", long_df.shape)
print(long_df.head(10))
```

 Tidy Data prinsibinə uyğun olaraq, illəri sütun şəklindən sətirlərə (melt funksiyası vasitəsilə) keçirdik. Bu, gələcəkdə qrafiklər çəkməyi və analizi çox asanlaşdıracaq.

```python
# Mətni rəqəmə çevirən funksiya (vergülü nöqtə ilə əvəz edir, xətaları NaN edir)
def to_number(s):
    s = (s.astype(str).str.replace(" ", "", regex=False)
         .str.replace(",", ".", regex=False))
    return pd.to_numeric(s, errors="coerce")

# 'value' sütununu rəqəm tipinə çevririk
long_df["value"] = to_number(long_df["value"])

print("Rəqəm çevrilməsi tamamlandı!")
print(long_df.info())
```

 Excel-dən gələn rəqəmlər daxildə mətn kimi qala bilər və ya içində boşluqlar ola bilər. Bu funksiya həmin simvolları təmizləyərək analizmiz üçün yararlı rəqəm formatına (float və ya int) çevirir

```python
def validate_data(df):
    # İllərin məntiqi aralıqda olmasını yoxlayırıq
    # (year sütununu əvvəl ədədi saata çevirmək lazımdır)
    df["year_clean"] = pd.to_numeric(df["year"].astype(str).str.extract(r'(\d{4})')[0], errors="coerce")

    # Mənfi dəyərin olub-olmamasını yoxlayırıq
    assert (df["value"].dropna() >= 0).all(), "Diqqət: Mənfi dəyər aşkarlandı!"

    print("Validasiya uğurla keçdi! Data analizə və qrafiklərə tam hazırdır.")
    return df

long_df = validate_data(long_df)
print(long_df.head(10))
```

 validate() funksiyası datada məntiqi səhvlərin (məsələn, mənfi rəqəmlər və ya səhv illər) olub-olmadığını avtomatik yoxlayır və kodun təhlükəsiz işləməsini təmin edir

```python
import pandas as pd
import numpy as np

file_path = '008 (4).xlsx'
df_raw = pd.read_excel(file_path, header=None, sheet_name='QUBA-XAÇMAZ')

all_data = []
current_section = "Ümumi"

# Cədvəldə sətirləri gəzərək bölmələri və illik göstəriciləri toplayırıq
i = 0
while i < len(df_raw):
    row = df_raw.iloc[i]
    val_0 = str(row[0]).strip()
    val_1 = str(row[1]).strip()

    # Bölmə başlığını yoxlayırıq
    if pd.notna(row[1]) and (any(word in val_1 for word in ['GÖSTƏRİCİLƏR', 'SƏHİYYƏ', 'TƏHSİL', 'İQTİSADİYYat'])):
        current_section = val_1
        i += 1
        continue

    # İllərin yerləşdiyi sətri tapırıq (məsələn: 2015, 2020...)
    if any(str(x).strip() in ['2015', '2020', '2021'] for x in row.values):
        years = row.values[2:8]  # İllər sütunları
        i += 1
        # Həmin bölmənin göstəricilərini oxuyuruq
        while i < len(df_raw):
            ind_row = df_raw.iloc[i]
            ind_name = ind_row[1]

            # Əgər sətir boşdursa və ya yeni bölmə başşlayırsa dövrdən çıxırıq
            if pd.isna(ind_name) or str(ind_name).strip() == '':
                break
            if any(word in str(ind_name) for word in ['GÖSTƏRİCİLƏR', 'SƏHİYYƏ', 'TƏHSİL']):
                i -= 1
                break

            values = ind_row.values[2:8]
            for y, v in zip(years, values):
                all_data.append({
                    "section": current_section,
                    "indicator": ind_name,
                    "year": y,
                    "value": v
                })
            i += 1
    else:
        i += 1

# Toplanan məlumatları DataFrame-ə çeviririk
long_df = pd.DataFrame(all_data)

# Şərti simvolları NaN edirik və rəqəmə çevirik
long_df["value"] = long_df["value"].astype(str).str.replace(" ", "", regex=False).str.replace(",", ".", regex=False)
long_df["value"] = pd.to_numeric(long_df["value"], errors="coerce")

# Boş dəyərləri təmizləyirik
long_df = long_df.dropna(subset=["value"])
long_df["year_clean"] = pd.to_numeric(long_df["year"], errors="coerce")

print("Datanın uğurla doldurulmuş ölçüsü (sətir, sütun):", long_df.shape)
print(long_df.head(10))

# CSV faylı olaraq yadda saxlayırıq
output_file = 'quba_xacmaz_clean_data.csv'
long_df.to_csv(output_file, index=False, encoding='utf-8-sig')
print("\nCSV faylı artıq tam doludur və uğurla yadda saxlanıldı!")
```
